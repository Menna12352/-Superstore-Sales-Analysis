```python
df[['Profit','Sales','Profit Margin']]
```

```python
df['Shipping Duration']=(df['Ship Date'] -df['Order Date'])
```

```python
df['Shipping Duration']=df['Shipping Duration'].dt.days
```

```python
df[['Ship Date','Order Date','Shipping Duration']]
```

```python
sales_mean = df["Sales"].mean()
print(sales_mean)
```

```python
df["Sales Performance Category"] = np.where(df["Sales"] > sales_mean, "High", "Normal")
```

```python
df[['Sales','Sales Performance Category']]
```

```python
df['Discount_Category'] = np.where(df['Discount'] > 0, 'Discounted', 'No Discount')
```

```python
df[['Discount','Discount_Category']]
```

```python
df['Profit_Category'] = np.where(
    df['Profit'] >= 100, 
    'High Profit', 
    np.where(df['Profit'] >= 0, 'Medium Profit', 'Loss')
)
```

```python
df[['Profit','Profit_Category']]
```

```python
df['Order_type']=np.where(
    df['Quantity'] >= 8, 
    'Large Order', 
    np.where(df['Quantity'] >= 4, 'Medium Order', 'Small Order')
)
```

```python
df[['Quantity','Order_type']]
```
